### Homework Problem 3 (8 Marks)
**Source Task:** Convert temperature from Celsius to Fahrenheit, and from Fahrenheit to Celsius.

**Authoritative Formulas from PDF:**
- Celsius to Fahrenheit: `F = (9 / 5) * C + 32`
- Fahrenheit to Celsius: `C = (5 / 9) * (F - 32)`

*(Note: Floating-point precision will be evaluated with standard decimal tolerance).*

**Requirements:**
- Given `celsius_input = 25`, compute `fahrenheit_result` (4 marks).
- Given `fahrenheit_input = 77`, compute `celsius_result` (4 marks).
- Print both results.

```python
# YOUR CODE HERE for Problem 3
celsius_input = 25
fahrenheit_result = (9 / 5) * celsius_input + 32

fahrenheit_input = 77
celsius_result = (5 / 9) * (fahrenheit_input - 32)

print("Celsius to Fahrenheit:", fahrenheit_result)
print("Fahrenheit to Celsius:", celsius_result)
```

### Homework Problem 4 (8 Marks)
**Source Task:** Write a Python program that will take a 4-digit number and print the **leftmost digit** of that number.

**Hint from Source:** Use floor division (`//`).

**Requirements:**
- Given a 4-digit number `num_4digit = 3564` (or any positive 4-digit integer):

- For example: if `num_4digit = 3564`, `leftmost_digit` must be `3`.
