### 1. if and else Statements
The `if` statement executes a code block when a condition evaluates to `True`. The `else` statement provides an alternative path when the condition evaluates to `False`.

```python
age = 18
if age >= 18:
    print('You are eligible to vote.')
else:
    print('You are not eligible to vote.')
```

### 2. elif Statement
Use `elif` (else-if) to check multiple conditions sequentially. Only the first condition that evaluates to `True` runs.

```python
score = 85
if score >= 90:
    print('Grade: A')
elif score >= 80:
    print('Grade: B')
else:
    print('Grade: C')
```

### 3. Nested if-else Statements
An `if` statement placed inside another `if` block allows multi-level decision making.

```python
age = 20
has_ticket = True
if age >= 18:
    if has_ticket:
        print('You are allowed to enter the event.')
    else:
        print('You need a ticket to enter.')
else:
    print('You are too young to enter.')
```

### 4. while Loops
A `while` loop repeats code as long as its condition remains `True`. Remember to update the loop variable to avoid an infinite loop!

```python
x = 5
while x > 0:
    print('Countdown:', x)
    x -= 1
print('Blast off!')
```

### 6. The range() Function
`range(stop)` generates numbers from `0` to `stop - 1`.
`range(start, stop)` generates from `start` to `stop - 1`.
`range(start, stop, step)` generates increments of `step`.

```python
print('range(5):', list(range(5)))
print('range(2, 6):', list(range(2, 6)))
print('range(1, 10, 2):', list(range(1, 10, 2)))
```

### 7. Conditions Inside Loops

```python
for i in range(1, 11):
    if i % 3 == 0:
        print(f'{i} is a multiple of 3')
    else:
        print(f'{i} is not a multiple of 3')
```

### 8. Nested Loops
A nested loop is a loop inside another loop. The inner loop executes completely for each iteration of the outer loop.

```python
for i in range(1, 4):
    for j in range(1, 3):
        print(f'i={i}, j={j}')
```

### 9. break Statement
`break` immediately terminates the enclosing loop and resumes execution after the loop.

```python
for num in range(1, 10):
    if num == 5:
        print('Breaking loop at 5!')
        break
    print(num)
```

### 10. continue Statement
`continue` skips the rest of the current iteration and jumps to the next iteration of the loop.

```python
for num in range(1, 10):
    if num % 2 == 0:
        continue
    print('Odd number:', num)
```

### 11. pass Statement

### 12. Indentation Practice
In Python, 4 spaces of indentation define code blocks. Inconsistent or missing indentation causes an `IndentationError`.

```python
# Correct indentation example
for i in range(3):
    print('Outer:', i)
    for j in range(2):
        print('  Inner:', j)
```
