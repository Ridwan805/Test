## 3. Type Conversion

Convert between types using `int()`, `float()`, `str()`, and `bool()`.

```python
num_str = "150"
converted_int = int(num_str)
converted_float = float(converted_int)
converted_bool = bool(converted_int)

print("Original string:", num_str, type(num_str))
print("Converted integer:", converted_int, type(converted_int))
print("Converted float:", converted_float, type(converted_float))
print("Converted boolean:", converted_bool, type(converted_bool))
```

## 5. String Indexing and Slicing

Strings use 0-based indexing from the left and negative indexing from the right.

```python
greeting = "Hello, World!"
print("First character [0]:", greeting[0])
print("Last character [-1]:", greeting[-1])
print("Substring [0:5]:", greeting[0:5])
print("Every second char [::2]:", greeting[::2])
print("Reversed string [::-1]:", greeting[::-1])
```

## 8. Arithmetic Operators

Python provides `+`, `-`, `*`, `/`, `%` (modulus), `//` (floor division), and `**` (exponentiation).

```python
a, b = 17, 5
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("True Division:", a / b)
print("Floor Division:", a // b)
print("Modulus (Remainder):", a % b)
print("Exponentiation:", a ** 2)
```

## 9. Relational and Logical Operators

Relational operators (`==`, `!=`, `<`, `>`, `<=`, `>=`) and logical operators (`and`, `or`, `not`).

```python
x, y = 15, 20
print("x < y:", x < y)
print("x == y:", x == y)
print("x < y and y > 10:", x < y and y > 10)
print("x > 50 or y == 20:", x > 50 or y == 20)
print("not (x == y):", not (x == y))
```
