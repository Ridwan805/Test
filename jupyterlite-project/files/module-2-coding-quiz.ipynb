### Question 1: Input and Type Conversion (3 Marks)
**Task:**
- You are given string representations of a product price and quantity: `price_text = "29.99"` and `quantity_text = "8"`.
- Convert `price_text` to a float and store in `price`.
- Convert `quantity_text` to an integer and store in `quantity`.
- Print both converted variables and verify their data types.

```python
# YOUR CODE HERE for Question 1
price_text = "29.99"
quantity_text = "8"

price = float(price_text)
quantity = int(quantity_text)

print("Price:", price, type(price))
print("Quantity:", quantity, type(quantity))
```

### Question 2: Arithmetic Operators (4 Marks)
**Task:**
Using the variables `price` and `quantity` from Question 1, calculate:
1. `total_cost = price * quantity` (1 mark)
2. `half_cost = total_cost / 2` (1 mark)
3. `whole_units_per_pack = quantity // 3` (1 mark)
4. `remaining_units = quantity % 3` (1 mark)

Print all four results.

```python
# YOUR CODE HERE for Question 2
total_cost = price * quantity
half_cost = total_cost / 2
whole_units_per_pack = quantity // 3
remaining_units = quantity % 3

print("Total Cost:", total_cost)
print("Half Cost:", half_cost)
print("Whole Units Per Pack:", whole_units_per_pack)
print("Remaining Units:", remaining_units)
```

### Question 3: Strings and Formatting (4 Marks)
**Task:**
Given:
- `first_name = "Ada"`
- `last_name = "Lovelace"`

Create:
1. `full_name`: concatenate the two names with exactly one space between them (e.g. `"Ada Lovelace"`).
2. `full_name_upper`: the uppercase version of `full_name` using `.upper()`.
3. `name_length`: the length of `full_name` using `len()`.

Print all three results.

```python
# YOUR CODE HERE for Question 3
first_name = "Ada"
last_name = "Lovelace"

full_name = first_name + " " + last_name
full_name_upper = full_name.upper()
name_length = len(full_name)

print("Full Name:", full_name)
print("Uppercase:", full_name_upper)
print("Length:", name_length)
```

### Question 4: String Indexing and Slicing (3 Marks)
**Task:**
Given word `word = "Algorithms"`:
1. `first_char`: extract the first character using index `0`.
2. `last_char`: extract the last character using negative index `-1`.
3. `first_three`: slice the first three characters using `[:3]`.
4. `reversed_word`: reverse the entire string using slice step `[::-1]`.

Print all four results.

```python
# YOUR CODE HERE for Question 4
word = "Algorithms"

first_char = word[0]
last_char = word[-1]
first_three = word[:3]
reversed_word = word[::-1]

print("First:", first_char)
print("Last:", last_char)
print("First Three:", first_three)
print("Reversed:", reversed_word)
```

### Question 5: Relational Operators (3 Marks)
**Task:**
Using `quantity` (from Q1) and `name_length` (from Q3):
1. `same_value = (quantity == name_length)` (1 mark)
2. `quantity_larger = (quantity > name_length)` (1 mark)
3. `different_value = (quantity != name_length)` (1 mark)

Print all three Boolean results.

```python
# YOUR CODE HERE for Question 5
same_value = (quantity == name_length)
quantity_larger = (quantity > name_length)
different_value = (quantity != name_length)

print("Same Value:", same_value)
print("Quantity Larger:", quantity_larger)
print("Different Value:", different_value)
```

### Question 6: Logical Operators (3 Marks)
**Task:**
Using `quantity` and `name_length`:
1. `logic_and = (quantity > 5 and name_length > 5)` (1 mark)
2. `logic_or = (quantity < 10 or name_length < 10)` (1 mark)
3. `logic_not = not (quantity == name_length)` (1 mark)

Print all three Boolean results.

```python
# YOUR CODE HERE for Question 6
logic_and = (quantity > 5 and name_length > 5)
logic_or = (quantity < 10 or name_length < 10)
logic_not = not (quantity == name_length)

print("Logic AND:", logic_and)
print("Logic OR:", logic_or)
print("Logic NOT:", logic_not)
```
