### 2. Indexing, Slicing, and Mutability
Try indexing positive and negative indices, slicing `[start:stop:step]`, and reversing with `[::-1]`.

```python
fruits = ["apple", "banana", "cherry", "date"]
print("First fruit:", fruits[0])
print("Last fruit:", fruits[-1])
print("Slice 1 to 3:", fruits[1:3])
print("Reversed:", fruits[::-1])

# Mutability check
fruits[1] = "blueberry"
print("Mutated list:", fruits)
```

### 4. List Comprehension
#### Practice Task 4.6: Odd Numbers 1 to 30
Generate a list of odd numbers between 1 and 30 using `[x for x in iterable if condition]`.

```python
# Task 4.6: List comprehension for odd numbers 1 to 30
odds = [x for x in range(1, 31) if x % 2 != 0]
print("Odd numbers:", odds)
```

### 7. Dictionaries: Access, Methods, and Updates
#### Practice Tasks 4.13B & 4.15

```python
# Task 4.13B: Display names with marks > 70
marks = {'Alex': 87, 'Ben': 89, 'Ross': 78, 'Ted': 66, 'Phil': 68, 'Cam': 90}
high_marks = [name for name, score in marks.items() if score > 70]
print("Marks > 70:", high_marks)

# Task 4.15: Merge two dictionaries
d1 = {'Harry': 15, 'Draco': 8, 'Nevil': 19}
d2 = {'Ginie': 18, 'Luna': 14}
merged = d1.copy()
merged.update(d2)
print("Merged dictionary:", merged)
```
